```python

# ==========================================
#       PERSONAL BUDGET TRACKER
#       Using Pandas and NumPy
# ==========================================

import pandas as pd
import numpy as np

# Read existing dataset
data = pd.read_csv("transactions.csv")

while True:

    print("\n================================")
    print("       MY BUDGET TRACKER")
    print("================================")

    print("\n1. View All Transactions")
    print("2. Add New Transaction")
    print("3. View Budget Summary")
    print("4. View Expenses By Category")
    print("5. Exit")

    choice = input("\nEnter your choice (1-5): ")

    # OPTION 1: VIEW DATASET
    if choice == "1":

        print("\nYOUR TRANSACTION DATA:")
        print(data)

    # OPTION 2: ADD NEW TRANSACTION
    elif choice == "2":

        print("\nENTER NEW TRANSACTION DETAILS")

        date = input("Enter date (YYYY-MM-DD): ")
        description = input("Enter description: ")
        category = input("Enter category: ")

        transaction_type = input("Enter type (Income/Expense): ")
        transaction_type = transaction_type.capitalize()

        while transaction_type not in ["Income", "Expense"]:
            print("Invalid type! Enter Income or Expense.")
            transaction_type = input("Enter type (Income/Expense): ")
            transaction_type = transaction_type.capitalize()

        amount = float(input("Enter amount: "))

        # Create new transaction
        new_transaction = pd.DataFrame({
            "Date": [date],
            "Description": [description],
            "Category": [category],
            "Type": [transaction_type],
            "Amount": [amount]
        })

        # Add transaction to existing dataset
        data = pd.concat([data, new_transaction], ignore_index=True)

        # Save updated dataset
        data.to_csv("transactions.csv", index=False)

        print("\nTransaction added successfully!")

    # OPTION 3: BUDGET SUMMARY
    elif choice == "3":

        income = data[data["Type"] == "Income"]["Amount"]
        expenses = data[data["Type"] == "Expense"]["Amount"]

        total_income = np.sum(income)
        total_expenses = np.sum(expenses)

        balance = total_income - total_expenses

        print("\n========== BUDGET SUMMARY ==========")

        print("Total Income: Rs.", total_income)
        print("Total Expenses: Rs.", total_expenses)
        print("Remaining Balance: Rs.", balance)

        if len(expenses) > 0:
            print("Highest Expense: Rs.", np.max(expenses))
            print("Average Expense: Rs.", round(np.mean(expenses), 2))

    # OPTION 4: EXPENSES BY CATEGORY
    elif choice == "4":

        expense_data = data[data["Type"] == "Expense"]

        category_expenses = expense_data.groupby("Category")["Amount"].sum()

        print("\nEXPENSES BY CATEGORY:")
        print(category_expenses)

    # OPTION 5: EXIT
    elif choice == "5":

        print("\nThank you for using Budget Tracker!")
        break

    else:
        print("\nInvalid choice! Please select 1 to 5.")
```
